```python
def matrix_chain_order(p):
    """
    Find the minimum number of scalar multiplications
    required to multiply a chain of matrices.

    p[i-1] x p[i] is the dimension of matrix i.
    """

    n = len(p) - 1

    # dp[i][j] stores minimum multiplication cost
    dp = [[0] * (n + 1) for _ in range(n + 1)]

    # split[i][j] stores the position of the best split
    split = [[0] * (n + 1) for _ in range(n + 1)]

    # length is the chain length
    for length in range(2, n + 1):
        for i in range(1, n - length + 2):
            j = i + length - 1
            dp[i][j] = float('inf')

            # Try every possible split
            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i - 1] * p[k] * p[j]
                )

                if cost < dp[i][j]:
                    dp[i][j] = cost
                    split[i][j] = k

    return dp, split


def print_optimal_parentheses(split, i, j):
    """Print the optimal parenthesization."""
    if i == j:
        return f"A{i}"

    k = split[i][j]

    left = print_optimal_parentheses(split, i, k)
    right = print_optimal_parentheses(split, k + 1, j)

    return f"({left} × {right})"


# Example:
# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60

dimensions = [10, 30, 5, 60]

dp, split = matrix_chain_order(dimensions)

n = len(dimensions) - 1

print("Minimum number of scalar multiplications:", dp[1][n])
print("Optimal parenthesization:",
      print_optimal_parentheses(split, 1, n))
```
